### ChatOllama local LLM calls

Run model locally 



```python
from langchain_ollama import ChatOllama

llm = ChatOllama(model="ornith:9b", temperature=0.7)

llm.invoke("Who was the first person to go to the moon?")
```

### call the Groq model open source models 

```python
from langchain_groq import ChatGroq
import os
api_key = os.getenv("api_key")
llm = ChatGroq(model="openai/gpt-oss-120b", temperature=0.7, api_key=api_key)

llm.invoke("Who was the first person to go to the moon?")
```

```python
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate.from_template("write two line poem about {topic}")

#prompt_template.invoke({"topic": "cats"})
```

```python
llm = ChatGroq(model="openai/gpt-oss-120b", temperature=0.7, api_key=api_key)
llm
```

```python
chain = prompt_template | llm

response = chain.invoke({"topic": "cats"})
```

```python
print(response.content)
```
