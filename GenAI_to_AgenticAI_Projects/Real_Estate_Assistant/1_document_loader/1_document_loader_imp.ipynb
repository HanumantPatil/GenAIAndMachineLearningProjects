```python
from urllib.parse import urlsplit
from urllib.robotparser import RobotFileParser

import requests
from langchain_community.document_loaders import UnstructuredURLLoader

url = "https://www.cnbc.com/2026/05/20/nvidia-nvda-earnings-report-q1-2027.html"
user_agent = (
    "RealEstateAssistantBot/1.0 "
    "(https://github.com/HanumantPatil/GenAIAndMachineLearningProjects)"
)
headers = {"User-Agent": user_agent, "Accept-Encoding": "gzip"}

url_parts = urlsplit(url)
robots_url = f"{url_parts.scheme}://{url_parts.netloc}/robots.txt"
robots_response = requests.get(robots_url, headers=headers, timeout=30)
robots_response.raise_for_status()

robots = RobotFileParser()
robots.set_url(robots_url)
robots.parse(robots_response.text.splitlines())
if not robots.can_fetch(user_agent, url):
    raise PermissionError(f"robots.txt does not allow fetching {url}")

loader = UnstructuredURLLoader(
    urls=[url],
    headers=headers,
    continue_on_failure=False,
)
docs = loader.load()

len(docs)
```
