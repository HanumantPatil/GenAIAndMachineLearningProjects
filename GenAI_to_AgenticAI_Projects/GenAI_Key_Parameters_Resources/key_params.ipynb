### Temperature

```python
from transformers import pipeline
generator = pipeline('text-generation', model='gpt2')
generator("Here is a suggestion on my new coffee shop name. The name should be:", temperature=0.1)
```

```python
generator("Here is a suggestion on my new coffee shop name. The name should be:", temperature=1.5)
```

### Top-p

```python
# Top-p (Creative and Diverse Output)
generator("The cat sat on", max_new_tokens=3, top_p=0.2)
```

```python
# Top-p (Creative and Diverse Output)
generator("The cat sat on", max_new_tokens=3, top_p=0.9)
```

### Top-k

```python
# Top-k (Controlled, Predictable Output)
generator("The cat sat on", max_new_tokens=3, top_k=1)
```

```python
# Top-k (Controlled, Predictable Output)
generator("The cat sat on", max_new_tokens=3, top_k=20)
```

### Output Length

```python
generator("The cat sat on", max_length=6)
```
