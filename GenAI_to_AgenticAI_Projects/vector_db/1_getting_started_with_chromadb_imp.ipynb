```python
import chromadb
```

```python
client = chromadb.Client()
```

```python
collection = client.create_collection(name="news")
```

```python
# Created embeddings for the documents and added them to the collection

collection.add(
    documents=[
        "Apple reported a quarterly profit of $20 billion, exceeding expectations.",
        "The Apple has lot of vitamins and minerals which are good for health.",
        "Apple is looking to expand its market share in the smartphone industry.",
        "The new iPhone model has been released with several new features.",
    ],
    ids=["doc1", "doc2", "doc3", "doc4"],
)
```

```python
result = collection.query(
    query_texts=["New iPhone will launch in September."],
    n_results=2
)

print(result)
```

```python
result1 = collection.query(
    query_texts=["My niece loves oranges."],
    n_results=2
)

print(result1)
```

### Persistent client

```python
clientp = chromadb.PersistentClient(path="./news_vectors_db")
```

```python
from chromadb.utils import embedding_functions

ef = embedding_function = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-distilroberta-v1"
)

collection = clientp.create_collection(
    name="news_v2",
    embedding_function=ef
)
```

```python
# Created embeddings for the documents and added them to the collection

collection.add(
    documents=[
        "Apple reported a quarterly profit of $20 billion, exceeding expectations.",
        "The Apple has lot of vitamins and minerals which are good for health.",
        "Apple is looking to expand its market share in the smartphone industry.",
        "The new iPhone model has been released with several new features.",
    ],
    ids=["doc1", "doc2", "doc3", "doc4"],
)
```

```python
result2 = collection.query(
    query_texts=["My niece loves oranges."],
    n_results=2
)

print(result2)
```
