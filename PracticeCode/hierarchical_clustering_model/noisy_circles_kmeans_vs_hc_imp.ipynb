```python
import matplotlib.pyplot as plt
import numpy as np

from sklearn import datasets, cluster
from sklearn.preprocessing import StandardScaler

n_samples = 1500
X,y = datasets.make_circles(n_samples=n_samples, factor=0.5, noise=0.05, random_state=170)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], s=10)
plt.title("Noisy circles dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=2, random_state=0)
kmeans.fit(X)
kmeans_labels = kmeans.labels_
```

```python
plt.scatter(X[:, 0], X[:, 1], c=kmeans_labels, edgecolor="k")
plt.title("KMeans Clustering on Noisy Circles")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.cluster import AgglomerativeClustering

agglo = AgglomerativeClustering(n_clusters=2, linkage="single")
agglo_labels = agglo.fit_predict(X)
plt.scatter(X[:, 0], X[:, 1], c=agglo_labels, edgecolor="k")
plt.title("Agglomerative Clustering (Single Linkage) on Noisy Circles")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
