```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split

X, y = make_classification(
    n_features=10, 
    n_samples=1000, 
    n_informative=8,
    n_redundant=2,
    n_repeated=0,
    n_classes=2, 
    random_state=42
)
```

### Method 01: Evaluate the model using train, test split and tune parameters by trail and error

```python
from sklearn.metrics import classification_report
from sklearn.tree import DecisionTreeClassifier

# Split data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

model = DecisionTreeClassifier(criterion="gini", max_depth=10) # criteria: "gini" or "entropy", max_depth=5 or 10
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
report = classification_report(y_test, y_pred)
print(report)
```

### Method 2: Cross val score

```python
from sklearn.model_selection import cross_val_score
model = DecisionTreeClassifier(criterion="gini", max_depth=5)
cross_val_scores = cross_val_score(model, X, y, cv=5)
print("Cross-validation scores:", cross_val_scores)
print("Average cross-validation score:", cross_val_scores.mean())
```

```python
from sklearn.model_selection import cross_val_score
model = DecisionTreeClassifier(criterion="gini", max_depth=10)
cross_val_scores = cross_val_score(model, X, y, cv=5)
print("Cross-validation scores:", cross_val_scores)
print("Average cross-validation score:", cross_val_scores.mean())
```

```python
from sklearn.model_selection import cross_val_score
model = DecisionTreeClassifier(criterion="entropy", max_depth=5)
cross_val_scores = cross_val_score(model, X, y, cv=5)
print("Cross-validation scores:", cross_val_scores)
print("Average cross-validation score:", cross_val_scores.mean())
```

```python
criterion_options = ["gini", "entropy"]
max_depth_options = [5, 10, 15]
for criterion in criterion_options:
    for max_depth in max_depth_options:
        model = DecisionTreeClassifier(criterion=criterion, max_depth=max_depth)
        cross_val_scores = cross_val_score(model, X, y, cv=5)
        print(f"Criterion: {criterion}, Max Depth: {max_depth}")
        print("Cross-validation scores:", cross_val_scores)
        print("Average cross-validation score:", cross_val_scores.mean())
        print("-" * 50)
```

```python
from sklearn.model_selection import GridSearchCV
param_grid = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [5, 10, 15]
}
grid_search = GridSearchCV(estimator=DecisionTreeClassifier(), param_grid=param_grid, cv=5, return_train_score=False)
grid_search.fit(X, y)
print("Best parameters:", grid_search.best_params_)
```
